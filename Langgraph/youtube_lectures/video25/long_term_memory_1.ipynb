```python
store = InMemoryStore(index={'embed':embedding_model,'dims':1536})
```

```python
namespace = ('users','u1')
store.put(namespace,'1',{'data':'i prefer very small line answer'})
store.put(namespace,'2',{'data':'i prefer concise answer'})
store.put(namespace,'3',{'data':'i prefer answer in rude way'})
store.put(namespace,'4',{'data':'i like coke'})
store.put(namespace,'5',{'data':'i like oranges'})
```

```python
items = store.search(namespace,query='what the user usually eat',limit=2)
for i in items:
    print(i.value)
```
