```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("datasets/diabetes.csv")
```

```python
df.head()
```

```python
df.corr()['Outcome']
```

```python
x = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=1)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(32,activation='relu',input_dim=8))
model.add(Dense(1,activation='sigmoid'))

model.compile(optimizer='Adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

```python
history = model.fit(x_train,y_train,batch_size=32,epochs=10,validation_data=(x_test,y_test))
```

```python
# how to select appropriate optimizer 
# no. of nodes in the layer
# no. of layers
# All in one model 
```

```python
import keras_tuner as kt
```

```python
def build_model(hp):
    model = Sequential()
    model.add(Dense(32,activation='relu',input_dim=8))
    model.add(Dense(1,activation='sigmoid'))

    optimizer = hp.Choice('optimizer', values = ['adam','sgd','rmsprop','adadelta'])

    model.compile(optimizer=optimizer, loss='binary_crossentropy' ,metrics=['accuracy'])
    return model

```

```python
tuner = kt.RandomSearch(build_model, objective='val_accuracy', max_trials=5)
```

```python
tuner.search(x_train,y_train, epochs=5,validation_data=(x_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
def build_model(hp):
    model = Sequential()
    
    units = hp.Int('units',min_value=8,max_value=128,step=8)

    model.add(Dense(units=units,activation='relu',input_dim=8))
    model.add(Dense(1,activation='sigmoid'))

    model.compile(optimizer='adam', loss='binary_crossentropy' ,metrics=['accuracy'])
    return model
```

```python
tuner = kt.RandomSearch(build_model, objective='val_accuracy', max_trials=5)
```

```python
tuner.search(x_train,y_train, epochs=5,validation_data=(x_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
def build_model(hp):
    model = Sequential()
    
    model.add(Dense(64,activation='relu',input_dim=8))
    
    for i in range(hp.Int('num_layers',min_value=1,max_value=100)):
        model.add(Dense(64,activation='relu'))
        
    model.add(Dense(1,activation='sigmoid'))

    model.compile(optimizer='adam', loss='binary_crossentropy' ,metrics=['accuracy'])
    return model
```

```python
tuner = kt.RandomSearch(build_model, objective='val_accuracy', max_trials=5)
```

```python
tuner.search(x_train,y_train, epochs=5,validation_data=(x_test,y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
def build_model(hp):
    model = Sequential()
    
    counter = 0

    for i in range(hp.Int('num_layers',min_value=1,max_value=100)):
        if(counter == 0):
            model.add(
                Dense(
                    units=hp.Int('units',min_value=1,max_value=100),
                    activation=hp.Choice('activation'+str(i),values=['tanh','sigmoid','relu']),
                    input_dim=8
                )
            )
        else:
            model.add(
                Dense(
                    units=hp.Int('units',min_value=1,max_value=100),
                    activation=hp.Choice('activation'+str(i),values=['tanh','sigmoid','relu'])
                )
            )
        counter += 1

    model.add(Dense(1,activation='sigmoid'))

    model.compile(
        optimizer=hp.Choice('optimizer',values=['adam','sgd','rmsprop','adadelta','nadam']), 
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    return model
```

```python
tuner = kt.RandomSearch(build_model, objective='val_accuracy', max_trials=3)
```

```python
tuner.search(x_train,y_train, epochs=5,validation_data=(x_test,y_test))
```
