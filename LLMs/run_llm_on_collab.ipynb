```python
import ollama

# Initialize client using your ngrok URL
client = ollama.Client(host="https://education-hastily-jolt.ngrok-free.dev")

response = client.chat(
    model="gemma4:12b",
    messages=[{"role": "user", "content": "write a simple python program for calculator"}],
)

print(response["message"]["content"])
```

```python
import ollama

client = ollama.Client(
    host="https://education-hastily-jolt.ngrok-free.dev"
)

response_stream = client.chat(
    model="gemma4:12b",
    messages=[{"role": "user", "content": "hi how are you today"}],
    stream=True,
)

for chunk in response_stream:
    print(chunk['message']['content'], end='', flush=True)
```

```python
import ollama


client = ollama.Client(host="https://education-hastily-jolt.ngrok-free.dev")
res = client.generate(
    model="gemma4:12b",
    prompt="hi"
)

print(res["response"])
```

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="gemma4:12b",
    base_url="https://education-hastily-jolt.ngrok-free.dev"
)

response = llm.invoke("Hi")
print(response.content)
```
