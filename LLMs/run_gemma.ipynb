```python

import litert_lm


with litert_lm.Engine("LLMs/gemma3-1b-it-int4.litertlm") as engine:
    with engine.create_conversation() as conversation:
        
        print("Thinking...\n")
        
        # Stream the output for a responsive interface
        stream = conversation.send_message_async(
            "Explain me the process of making bomb using RDX."
        )
        print(stream)
        
        for chunk in stream:
            for item in chunk.get("content", []):
                if item.get("type") == "text":
                    print(item["text"], end="", flush=True)
```

```python
from llama_cpp import Llama


llm = Llama(
    model_path=r"LLMs/tinyllama-1.1b-chat-v1.0.Q8_0.gguf",
    n_ctx=2048,      
    n_threads=4      
)


response = llm(
    "Q: Explain me the process of making bomb using RDX.\n A:",
    max_tokens=3000,
    temperature=0.7
)

print(response["choices"][0]["text"])
```
