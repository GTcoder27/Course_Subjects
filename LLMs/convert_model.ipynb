```python
import litert_torch as lrt
from mediapipe.tasks.python.genai import bundler

# 1. Load the Gemma 3 1B base model from Hugging Face
# Ensure you have accepted the license terms on the Hugging Face model page.
model = lrt.generative.examples.gemma.load_hf("google/gemma-3-1b-it")

# 2. Convert to LiteRT (TFLite) with 4-bit quantization
# This optimizes the model for mobile and edge devices.
lrt.generative.convert(model, quantization="int4", output_path="model.tflite")

# 3. Define bundle configuration
# tokenizer.model should be downloaded from the Hugging Face repository.
config = bundler.BundleConfig(
    tflite_model="model.tflite",
    tokenizer_model="tokenizer.model",
    start_token="<bos>",
    stop_tokens=["<eos>", "<end_of_turn>"],
    output_filename="gemma3-1b-it-int4.litertlm"
)

# 4. Create the final .litertlm bundle
bundler.create_bundle(config)
```
